# Explainable AI — SHAP on the Diagnosis Model

Loads the trained diagnosis model and generates real SHAP explanations for sample predictions.

```
import sys, joblib, shap, numpy as np
sys.path.insert(0, '../backend')
model = joblib.load('../backend/model/diagnosis_model.pkl')
scaler = joblib.load('../backend/model/diagnosis_scaler.pkl')
features = joblib.load('../backend/model/diagnosis_features.pkl')
import pandas as pd
efdb = pd.read_csv('../data/EngineFaultDB/EngineFaultDB_Final.csv')
row = efdb.sample(1, random_state=7)
x = scaler.transform(row[features].values)
explainer = shap.TreeExplainer(model)
sv = explainer.shap_values(x)
print('Predicted class:', model.predict(x)[0])
sv
```

```
shap.summary_plot(sv, x, feature_names=features, plot_type='bar', show=True) if hasattr(sv, 'shape') else print('multi-output SHAP array — see backend/ml_core.py run_diagnosis() for the per-class handling used in the live API')
```
