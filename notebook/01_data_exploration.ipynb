# Data Exploration — All Five Datasets

Inspects every dataset actually used in this project (see docs/dataset_sources.md). No schema is assumed — everything is read directly from disk.

```python
import pandas as pd, json, os
DATA = os.path.join('..', 'data')
```

## 1. EngineFaultDB (real, fault diagnosis)

```python
efdb = pd.read_csv(os.path.join(DATA, 'EngineFaultDB', 'EngineFaultDB_Final.csv'))
print(efdb.shape)
efdb.describe()
```

```python
efdb['Fault'].value_counts().sort_index()
```

## 2. NASA C-MAPSS FD001 (real, RUL — turbofan, NOT automotive)

```python
cols = ['unit','cycle','op1','op2','op3'] + [f'sensor_{i}' for i in range(1,22)]
cmapss = pd.read_csv(os.path.join(DATA, 'NASA_RUL', 'train_FD001.txt'), sep=r'\s+', header=None, names=cols)
print(cmapss.shape, cmapss['unit'].nunique(), 'units')
cmapss.groupby('unit').size().describe()
```

## 3. Automotive Faults knowledge base (real, RAG/maintenance layer)

```python
kb = json.load(open(os.path.join(DATA, 'Automotive_Faults', 'automotive_faults_aktc_obike_et_al.json')))
print(len(kb), 'entries')
pd.Series([e['category'] for e in kb]).value_counts()
```

## 4. Synthetic Demo Telemetry (NOT real vehicle data)

```python
demo = pd.read_csv(os.path.join(DATA, 'demo', 'demo_vehicle_telemetry.csv'))
print(demo.shape)
demo['vehicle_id'].value_counts()
```
